```python
import numpy as np
import pandas as pd

# -------------------------------
# DATA (EMBEDDED)
# -------------------------------
data = pd.DataFrame({
    "Model": ["BERT", "RoBERTa", "Sentence-BERT", "DistilBERT"],
    "CosineSimilarity": [0.87, 0.89, 0.91, 0.85],
    "Accuracy": [0.92, 0.94, 0.93, 0.90],
    "InferenceTime": [120, 150, 90, 60],
    "ModelSize": [420, 500, 350, 250]
})

# -------------------------------
# TOPSIS PARAMETERS
# -------------------------------
weights = np.array([0.4, 0.3, 0.2, 0.1])
impacts = ['+', '+', '-', '-']

models = data["Model"].values
decision_matrix = data.iloc[:, 1:].values.astype(float)

# -------------------------------
# STEP 1: NORMALIZATION
# -------------------------------
norm_matrix = decision_matrix / np.sqrt((decision_matrix ** 2).sum(axis=0))

# -------------------------------
# STEP 2: WEIGHTED NORMALIZATION
# -------------------------------
weighted_matrix = norm_matrix * weights

# -------------------------------
# STEP 3: IDEAL BEST & WORST
# -------------------------------
ideal_best = np.zeros(weighted_matrix.shape[1])
ideal_worst = np.zeros(weighted_matrix.shape[1])

for i in range(weighted_matrix.shape[1]):
    if impacts[i] == '+':
        ideal_best[i] = weighted_matrix[:, i].max()
        ideal_worst[i] = weighted_matrix[:, i].min()
    else:
        ideal_best[i] = weighted_matrix[:, i].min()
        ideal_worst[i] = weighted_matrix[:, i].max()

# -------------------------------
# STEP 4: DISTANCES
# -------------------------------
distance_best = np.sqrt(((weighted_matrix - ideal_best) ** 2).sum(axis=1))
distance_worst = np.sqrt(((weighted_matrix - ideal_worst) ** 2).sum(axis=1))

# -------------------------------
# STEP 5: TOPSIS SCORE
# -------------------------------
topsis_score = distance_worst / (distance_best + distance_worst)

# -------------------------------
# STEP 6: RANKING
# -------------------------------
result = pd.DataFrame({
    "Model": models,
    "TOPSIS Score": topsis_score
})

result["Rank"] = result["TOPSIS Score"].rank(ascending=False)
result = result.sort_values("Rank")

# -------------------------------
# SAVE CSV (SAME DIRECTORY)
# -------------------------------
result.to_csv("topsis_result.csv", index=False)

# -------------------------------
# DISPLAY
# -------------------------------
print("TOPSIS completed successfully!\n")
print(result)
print("\nCSV file saved as: topsis_result.csv")

plt.figure(figsize=(8, 5))
plt.bar(result["Model"], result["TOPSIS Score"])
plt.xlabel("Pre-trained Models")
plt.ylabel("TOPSIS Score")
plt.title("TOPSIS Ranking of Pre-trained Models")
plt.tight_layout()
plt.savefig("topsis_ranking.png")
plt.close()

# -------------------------------
# DISPLAY
# -------------------------------
print("TOPSIS completed successfully!\n")
print(result)
print("\nGenerated files:")
print("✔ topsis_result.csv")
print("✔ topsis_ranking.png")
```
